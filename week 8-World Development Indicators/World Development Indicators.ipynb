## PART 1

# Calculate annual growth

How much did services and merchandise exports increase per year

```python
trade = trade.rename(columns={
    'BX.GSR.NFSV.CD': 'Service_Exports',
    'BX.GSR.MRCH.CD': 'Goods_Exports'
})


# 1.Service export growth
trade['Service_Growth_%'] = (
    trade
    .groupby('Country Code')['Service_Exports']
    .pct_change() * 100
)

# 2.Merchandise export growth
trade['Goods_Growth_%'] = (
    trade
    .groupby('Country Code')['Goods_Exports']
    .pct_change() * 100
)
```

# Calculate total percentage growth

How much did services and merchandise exports increase for 24 years

```python

growth_summary = (
    trade
    .sort_values(['Country Code', 'Year'])
    .groupby('Country Code')
    .agg(
        Service_Start=('Service_Exports', 'first'),
        Service_End=('Service_Exports', 'last'),
        Goods_Start=('Goods_Exports', 'first'),
        Goods_End=('Goods_Exports', 'last')
    )
)
```

```python
growth_summary['Service_Total_Growth_%'] = (
    (growth_summary['Service_End'] -
     growth_summary['Service_Start'])
    / growth_summary['Service_Start']
) * 100

growth_summary['Goods_Total_Growth_%'] = (
    (growth_summary['Goods_End'] -
     growth_summary['Goods_Start'])
    / growth_summary['Goods_Start']
) * 100
```

```python
growth_summary
```

# Calculate CAGR

comparing growth over a long period from 2000 to 2024

```python
n_years = 2024 - 2000
```

```python
growth_summary['Service_CAGR_%'] = (
    (growth_summary['Service_End'] /
     growth_summary['Service_Start'])
    ** (1 / n_years) - 1
) * 100
```

```python
growth_summary['Goods_CAGR_%'] = (
    (growth_summary['Goods_End'] /
     growth_summary['Goods_Start'])
    ** (1 / n_years) - 1
) * 100
```

```python
growth_summary
```

 # Calculate the Service/Merchandise ratio

```python
trade['Service_Goods_Ratio'] = (
    trade['Service_Exports'] /
    trade['Goods_Exports']
)
```

```python
trade
```

```python
global_trade = (
    trade
    .groupby('Year')[['Service_Exports', 'Goods_Exports']]
    .sum(min_count=1)
    .reset_index()
)
```

```python
global_trade.head(25)
```

# Calculate global growth

```python
global_trade['Service_Growth_%'] = (
    global_trade['Service_Exports']
    .pct_change() * 100
)

global_trade['Goods_Growth_%'] = (
    global_trade['Goods_Exports']
    .pct_change() * 100
)
```

```python
global_trade['Service_Goods_Ratio'] = (
    global_trade['Service_Exports'] /
    global_trade['Goods_Exports']
)
```

# Calculate global CAGR

```python
service_cagr = (
    (global_trade['Service_Exports'].iloc[-1] /
     global_trade['Service_Exports'].iloc[0])
    ** (1 / 24) - 1
) * 100
```

```python
goods_cagr = (
    (global_trade['Goods_Exports'].iloc[-1] /
     global_trade['Goods_Exports'].iloc[0])
    ** (1 / 24) - 1
) * 100
```

```python
print(f"Service CAGR: {service_cagr:.2f}%")
print(f"Goods CAGR: {goods_cagr:.2f}%")
```

Between 2000 and 2024, global service exports grew at an estimated compound annual rate of 7.30%, compared with 6.34% for merchandise exports, indicating faster expansion of service exports over the period.

```python
composition_long['Service_Type'] = (
    composition_long['Indicator Code']
    .map({
        'BX.GSR.TRAN.ZS': 'Transport',
        'BX.GSR.TRVL.ZS': 'Travel',
        'BX.GSR.INSF.ZS': 'Insurance & Financial'
    })
)
```

```python
composition_long.head()
```

```python
missing = (
    composition_long
    .groupby('Service_Type')['Percentage']
    .apply(lambda x: x.isna().mean() * 100)
    .reset_index(name='Missing_%')
)

missing
```

Because these indicators are already percentages of service exports, we shouldn't sum them across countries.

Instead, we can calculate a global weighted share if we have the service-export dollar values from Part 1.

```python
trade
```

```python
composition = composition_long.merge(
    trade[
        ['Country Code', 'Year', 'Service_Exports']
    ],
    on=['Country Code', 'Year'],
    how='left'
)
```

###  Calculate the actual dollar value of each service category

```python
composition['Service_Value'] = (
    composition['Service_Exports'] *
    composition['Percentage'] / 100
)
```

### Calculate global totals

```python

global_composition = (
    composition
    .groupby(['Year', 'Service_Type'])['Service_Value']
    .sum(min_count=1)
    .reset_index()
)
```

### Calculate each category's global share

```python
# Calculate each category's global share
global_composition['Total_Three_Categories'] = (
    global_composition
    .groupby('Year')['Service_Value']
    .transform('sum')
)
```

```python
global_composition['Category_Share_%'] = (
    global_composition['Service_Value'] /
    global_composition['Total_Three_Categories']
) * 100
```

### Find the dominant service category

```python

dominant_service = (
    global_composition
    .sort_values(
        ['Year', 'Service_Value'],
        ascending=[True, False]
    )
    .groupby('Year')
    .first()
    .reset_index()
)

dominant_service[
    ['Year', 'Service_Type', 'Service_Value']
]
```

```python


fig = px.line(
    global_composition,
    x='Year',
    y='Category_Share_%',
    color='Service_Type',
    title='Composition of Global Service Exports',
    labels={
        'Category_Share_%': 'Share (%)',
        'Service_Type': 'Service Category'
    }
)

fig.show()
```

```python
fig = px.area(
    global_composition,
    x='Year',
    y='Category_Share_%',
    color='Service_Type',
    title='Changing Composition of Global Service Exports',
    labels={
        'Category_Share_%': 'Share of Selected Services (%)',
        'Service_Type': 'Service Category'
    },
    groupnorm='percent'
)

fig.show()
```

This lets you visually see whether:

* Travel became more important
* Transport declined
* Financial services increased
* The composition changed substantially around 2020

Analyze COVID-19 separately

```python
periods = pd.cut(
    global_composition['Year'],
    bins=[2014, 2019, 2021, 2024],
    labels=[
        'Pre-COVID',
        'COVID',
        'Post-COVID'
    ]
)

global_composition['Period'] = periods
```

```python
period_summary = (
    global_composition
    .groupby(['Period', 'Service_Type'], observed=True)
    ['Category_Share_%']
    .mean()
    .reset_index()
)

period_summary
```

```python
regions = [
    'EAS',   # East Asia & Pacific
    'ECS',   # Europe & Central Asia
    'LCN',   # Latin America & Caribbean
    'MEA',   # Middle East, North Africa, Afghanistan & Pakistan
    'NAC',   # North America
    'SAS',   # South Asia
    'SSF',   # Sub-Saharan Africa
    'AFE',  # Africa Eastern and Southern
    'AFW',  # Africa Western and Central
    'CEB',  # Central Europe and the Baltics
]
```

```python
regional_trade = trade[
    trade['Country Code'].isin(regions)
].copy()
```

### Calculate annual growth by region

```python
regional_trade = regional_trade.sort_values(
    ['Country Code', 'Year']
)

regional_trade['Service_Growth_%'] = (
    regional_trade
    .groupby('Country Code')['Service_Exports']
    .pct_change() * 100
)
```

### Calculate total growth from 2000–2024

```python
regional_growth = (
    regional_trade
    .sort_values(['Country Code', 'Year'])
    .groupby('Country Code')
    .agg(
        Start=('Service_Exports', 'first'),
        End=('Service_Exports', 'last')
    )
    .reset_index()
)
```

```python
regional_growth['Total_Growth_%'] = (
    (regional_growth['End'] - regional_growth['Start'])
    / regional_growth['Start']
) * 100
```

```python
regional_growth.sort_values(
    'Total_Growth_%',
    ascending=False
)
```

### Calculate regional CAGR

```python
regional_growth['CAGR_%'] = (
    (regional_growth['End'] / regional_growth['Start'])
    ** (1 / 24) - 1
) * 100
```

```python
regional_growth.sort_values(
    'CAGR_%',
    ascending=False
)
```

```python
import plotly.express as px

fig = px.bar(
    regional_growth.sort_values('CAGR_%'),
    x='CAGR_%',
    y='Country Code',
    orientation='h',
    title='Service Export CAGR by Region, 2000–2024',
    labels={
        'CAGR_%': 'CAGR (%)',
        'Country Code': 'Region'
    }
)

fig.show()
```

```python
regional_trade['Goods_Growth_%'] = (
    regional_trade
    .groupby('Country Code')['Goods_Exports']
    .pct_change() * 100
)
```

```python
regional_cagr = (
    regional_trade
    .sort_values(['Country Code', 'Year'])
    .groupby('Country Code')
    .agg(
        Service_Start=('Service_Exports', 'first'),
        Service_End=('Service_Exports', 'last'),
        Goods_Start=('Goods_Exports', 'first'),
        Goods_End=('Goods_Exports', 'last')
    )
    .reset_index()
)
```

```python
regional_cagr['Service_CAGR_%'] = (
    (regional_cagr['Service_End'] /
     regional_cagr['Service_Start'])
    ** (1 / 24) - 1
) * 100

regional_cagr['Goods_CAGR_%'] = (
    (regional_cagr['Goods_End'] /
     regional_cagr['Goods_Start'])
    ** (1 / 24) - 1
) * 100
```

```python
regional_cagr['CAGR_Difference'] = (
    regional_cagr['Service_CAGR_%'] -regional_cagr['Goods_CAGR_%']
)
```

Findings: SAS at 13.42% CAGR is the leading,followed by CEB at 9.15%, then NAC at 5.84%.

The COVID analysis is also strong: travel's selected-category share fell from 47.75% pre-COVID to 26.77% during COVID, while transport increased from 32.78% to 43.67%. Post-COVID, travel recovered to 40.28%.
